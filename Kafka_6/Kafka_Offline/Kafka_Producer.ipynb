```python
# prompt: give a code to provide kafka producer conf

import pandas as pd
import json
from kafka import KafkaProducer

producer = KafkaProducer(bootstrap_servers='localhost:9092')
```

```python
topic = 'ecommerce'

with open('customers.json','r') as file:
  customers_data = json.load(file)

value = customers_data[0]
key = value['customer_id']

print(key,value)
```

```python
type(str(value).encode('utf-8'))
```

```python
producer.send(
    topic,
    key=str(key).encode('utf-8'),
    value=json.dumps(value).encode('utf-8')
)
producer.flush()
```

```python

topic = 'ecommerce'

# with open('customers.json','r') as file:
#   customers_data = json.load(file)

sent_count = 0
failed_count = 0

for record in customers_data:
  try:

    message_value = json.dumps(record).encode('utf-8')
    message_key = str(int(record['customer_id']) + 1000).encode('utf-8')

    metadata = producer.send(topic, key=message_key, value=message_value).get(timeout=10)
    print(f"Message delivered to {metadata.topic} [{metadata.partition}] at offset {metadata.offset}")
    sent_count += 1

  except Exception as e:
    failed_count += 1
    print(f"Error sending message: {e}")

producer.flush()

print(f"Kafka send complete: {sent_count} sent, {failed_count} failed")
```
