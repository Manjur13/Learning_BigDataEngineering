```python
# Read Data from Via Locally Running Kafka Consumer
```

```python
import json
from kafka import KafkaConsumer

topic = 'ecommerce'
bootstrap_servers = 'localhost:9092'
consumer_group = 'ecommerce-consumer-group'
```

```python
def decode_message(message):
    key = message.key.decode('utf-8') if message.key is not None else None
    value = json.loads(message.value.decode('utf-8'))
    return key, value
```

```python
consumer = KafkaConsumer(
    topic,
    bootstrap_servers=bootstrap_servers,
    group_id=consumer_group,
    auto_offset_reset='earliest',
    enable_auto_commit=True,
)

print(f"Listening for messages on '{topic}'. Stop the cell to exit.")
try:
    while True:
        records = consumer.poll(timeout_ms=1000, max_records=100)
        for messages in records.values():
            for message in messages:
                key, value = decode_message(message)
                print(
                    f"Received {message.topic}[{message.partition}] "
                    f"offset={message.offset} key={key} value={value}"
                )
except KeyboardInterrupt:
    print("Consumer stopped.")
finally:
    consumer.close()
```
